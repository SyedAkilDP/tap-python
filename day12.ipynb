```
import csv

def show_source(file_name):
    print("Reading file:", file_name)

def read_csv(path):
    with open(path, mode="r", encoding="utf-8-sig", newline="") as file:
        reader = csv.DictReader(file)
        return list(reader)
```

```
from reader import read_csv

orders = read_csv(r"C:\sql\orders.csv")

print("Total orders:", len(orders))
print("First order:", orders[0])
```

```
from transformer import to_snake_case

print(to_snake_case("Customer ID"))
print(to_snake_case("OrderDate"))
print(to_snake_case("Product-Price"))
print(to_snake_case("Total Revenue"))
```

```
clean_orders = [
    {to_snake_case(key): value for key, value in order.items()}
    for order in orders
]

print("Original columns:", list(orders[0].keys()))
print("Cleaned columns:", list(clean_orders[0].keys()))
```

```
column_mapping = {
    "order_status": "status",
    "order_purchase_timestamp": "purchase_timestamp",
    "order_estimated_delivery_date": "estimated_delivery_date"
}

mapped_orders = [
    {
        column_mapping.get(key, key): value
        for key, value in order.items()
    }
    for order in clean_orders
]

print("Before:", list(clean_orders[0].keys()))
print("After:", list(mapped_orders[0].keys()))
```

```
from datetime import datetime

converted_orders = []

for order in mapped_orders:
    new_order = order.copy()

    purchase_date = new_order["purchase_timestamp"]

    if purchase_date:
        new_order["purchase_timestamp"] = datetime.strptime(
            purchase_date, "%Y-%m-%d %H:%M:%S"
        )
    else:
        new_order["purchase_timestamp"] = None

    converted_orders.append(new_order)

print(converted_orders[0]["purchase_timestamp"])
print(type(converted_orders[0]["purchase_timestamp"]))
```

```
from reader import read_csv

order_items = read_csv(r"C:\sql\order_items.csv")

converted_items = []

for item in order_items:
    new_item = item.copy()

    for column in ["price", "freight_value"]:
        value = new_item[column].strip()

        new_item[column] = float(value) if value else None

    converted_items.append(new_item)

print("Price:", converted_items[0]["price"])
print("Price type:", type(converted_items[0]["price"]))

print("Freight:", converted_items[0]["freight_value"])
print("Freight type:", type(converted_items[0]["freight_value"]))
```

```
import pandas as pd

df = pd.read_csv(r"C:\sql\order_items.csv")

df["item_total"] = df["price"] + df["freight_value"]

order_totals = df.groupby("order_id", as_index=False)["item_total"].sum()

print(order_totals.head())
```

```
order_totals_dict = {}

for item in df.to_dict("records"):
    order_id = item["order_id"]
    total = item["price"] + item["freight_value"]

    order_totals_dict[order_id] = order_totals_dict.get(order_id, 0) + total

print(list(order_totals_dict.items())[:5])
```
