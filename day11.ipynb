```python
import csv

file_path = r"C:\sql\orders.csv"

missing_values = {}

with open(file_path, "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    for column in reader.fieldnames:
        missing_values[column] = 0

    for row in reader:
        for column, value in row.items():
            if value is None or value.strip() == "":
                missing_values[column] += 1

print("Missing Values Per Column:")
for column, count in missing_values.items():
    print(column, ":", count)
```

```python
import csv

file_path = r"C:\sql\orders.csv"

with open(file_path, "r", encoding="utf-8-sig", newline="") as file:
    reader = csv.DictReader(file)

    missing_values = {column: 0 for column in reader.fieldnames}

    for row in reader:
        for column, value in row.items():
            if value is None or value.strip() == "":
                missing_values[column] += 1

print("Missing Values Per Column:")

for column, count in missing_values.items():
    print(f"{column}: {count}")
```

```python
from pathlib import Path

INPUT_FOLDER = Path(r"C:\sql")
```

```python
import csv

for file_path in sorted(INPUT_FOLDER.glob("*.csv")):
    with open(file_path, "r", encoding="utf-8-sig", newline="") as file:
        reader = csv.reader(file)

        columns = next(reader, [])
        row_count = sum(1 for row in reader)

        print("File Name:", file_path.name)
        print("Total Rows:", row_count)
        print("Total Columns:", len(columns))
        print("-------------------------")
```

```python
import csv
from pathlib import Path

INPUT_FOLDER = Path(r"C:\sql")

input_file = INPUT_FOLDER / "orders.csv"
output_file = INPUT_FOLDER / "orders_semicolon.csv"

with open(input_file, "r", encoding="utf-8-sig", newline="") as source:
    reader = csv.reader(source)

    with open(output_file, "w", encoding="utf-8", newline="") as target:
        writer = csv.writer(target, delimiter=";")
        writer.writerows(reader)

print("Semicolon CSV created successfully!")
```
